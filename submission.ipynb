Accepted post-IS operational version V4. Pure pricing definitions extracted from the official starter are in `strategy_defs.py`. Literal agreement tags in TOP100, prior-filing-session contracts, next-session close entry, one covered-call proxy. This broad pool does not assert buyer/control/first-public status. See README.md for execution and limitations. Original hypothesis/results remain in Git history and the ignored private archive.

All execution switches default to False. Contestant OOS requires explicit one-shot approval through the root runner. Judges can run their own supplied dates with `run_study(start, end, api_key, judges_window=True)`; no private SEC files or accession whitelist are required.


```python
from pathlib import Path
import sys
import json
import subprocess

PROJECT = Path.cwd()
if not (PROJECT / 'research').is_dir():
    raise RuntimeError('Open this notebook from the repository root.')
sys.path.insert(0, str(PROJECT / 'research'))
from submission_variant import run_submission_study as run_study

START_DATE, END_DATE = '2024-01-01', '2025-12-31'
ACCEPT_OPERATIONAL_REVISION = True  # adopted explicitly after original IS
RUN_REVISED_IS = False
RUN_CONTESTANT_OOS_ONCE = False
JUDGES_WINDOW = False  # judges alone may enable their own window
RUN_JUDGES_STUDY = False
```

```python
if RUN_REVISED_IS or RUN_CONTESTANT_OOS_ONCE:
    if not ACCEPT_OPERATIONAL_REVISION:
        raise RuntimeError('Review and explicitly adopt the disclosed event/timing revision first.')
    common = [sys.executable, str(PROJECT / 'run_research.py'), '--mode', 'submission',
              '--execute', '--accept-operational-revision']
    if RUN_REVISED_IS:
        subprocess.run(common + ['--start', '2024-01-01', '--end', '2025-12-31'], check=True)
    if RUN_CONTESTANT_OOS_ONCE:
        subprocess.run(common + ['--approve-oos-once', '--start', '2026-01-01', '--end', '2026-08-31'], check=True)
else:
    print('No research executed. Switches are off. OOS still requires explicit one-shot approval.')

if RUN_JUDGES_STUDY:
    if not JUDGES_WINDOW:
        raise RuntimeError('Only judges select the unseen-window mode.')
    # MASSIVE_API_KEY is loaded by the authenticating process. A judge may instead
    # pass their key as run_study(START_DATE, END_DATE, api_key, judges_window=True).
    study = run_study(START_DATE, END_DATE, judges_window=True)
    display(study.tables['theta_board'])
    display(study.tables['official_diff'])
    display(study.tables['implied_realized'])
```
